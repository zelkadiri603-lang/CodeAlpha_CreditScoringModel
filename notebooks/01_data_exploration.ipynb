```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Define the list of numerical features exhibiting right-skewness
skewed_features = ['credit_amount', 'duration_months', 'age', 'existing_credits']

# 2. Apply log transformation (log1p) and create new columns
for col in skewed_features:
    df[f'{col}_log'] = np.log1p(df[col])

# 3. Calculate and display skewness values before and after transformation
skewness_df = pd.DataFrame({
    'Original Skewness': df[skewed_features].skew(),
    'Log Transformed Skewness': df[[f'{col}_log' for col in skewed_features]].skew().values
})

print("--- Skewness Comparison ---")
print(skewness_df.round(3))

# 4. Plot distributions of the log-transformed features
plt.figure(figsize=(12, 8))
for i, col in enumerate(skewed_features, 1):
    plt.subplot(2, 2, i)
    sns.histplot(df[f'{col}_log'], kde=True, color='teal')
    plt.title(f'Distribution of {col} (Log Transformed)')

plt.tight_layout()
plt.show()
```

the distributions of credit_amount and age, significantly reducing right-skewness and outlier impact. duration_months displayed a multimodal structure representing standard bank loan durations, while existing_credits retained its discrete two-cluster behavior.

```python


# 1. Select all numerical features from the dataframe
df_numeric = df.select_dtypes(include=['int64', 'float64'])

# 2. Iterate through each numerical column to count IQR-based outliers
for col in df_numeric.columns:

    # Calculate First Quartile (Q1) and Third Quartile (Q3)
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    # Calculate Interquartile Range (IQR)
    IQR = Q3 - Q1

    # Define lower and upper bound thresholds
    minim = Q1 - 1.5 * IQR
    maxim = Q3 + 1.5 * IQR

    # Filter rows falling outside the lower and upper bounds
    outliers = df[(df[col] < minim) | (df[col] > maxim)]

    # Print summary count for each feature
    print(f"Feature: {col} | Outliers Count: {len(outliers)}")
```

```python

# 1. Define columns for outlier capping (using log-transformed features)
features_to_cap = ['credit_amount_log', 'duration_months_log', 'age_log']

# 2. Function to cap outliers based on IQR
def cap_outliers_IQR(df, columns):
    df_capped = df.copy()
    for col in columns:
        Q1 = df_capped[col].quantile(0.25)
        Q3 = df_capped[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        
        # Apply capping
        df_capped[col] = np.where(df_capped[col] > upper_bound, upper_bound,
                         np.where(df_capped[col] < lower_bound, lower_bound, df_capped[col]))
    return df_capped

# Apply capping
df_capped = cap_outliers_IQR(df, features_to_cap)

# 3. Visualize before and after capping
plt.figure(figsize=(14, 6))
for i, col in enumerate(features_to_cap, 1):
    plt.subplot(2, 3, i)
    sns.boxplot(y=df[col], color='skyblue')
    plt.title(f'Original: {col}')
    
    plt.subplot(2, 3, i + 3)
    sns.boxplot(y=df_capped[col], color='teal')
    plt.title(f'Capped: {col}')

plt.tight_layout()
plt.show()
```

```python


# 1. Calculate Monthly Payment Ratio in Log Scale
# log(A / B) = log(A) - log(B)
df['monthly_payment_log'] = df['credit_amount_log'] - df['duration_months_log']

# 2. Calculate Age-to-Duration Ratio (Financial Stability Index)
df['age_to_duration'] = df['age'] / df['duration_months']

# 3. Summary Statistics of Newly Created Features
print("--- Newly Created Features Summary ---")
print(df[['monthly_payment_log', 'age_to_duration']].describe().round(3))

# 4. Visualize Distributions and Target Relationships (Warning-Free Code)
plt.figure(figsize=(12, 5))

# Plot monthly_payment_log distribution vs Target
plt.subplot(1, 2, 1)
sns.boxplot(
    x='target',
    y='monthly_payment_log',
    data=df,
    hue='target',
    palette='Set2',
    legend=False,
)
plt.title('Monthly Payment (Log) vs Target')
plt.xlabel('Credit Risk Target')
plt.ylabel('Monthly Payment (Log)')

# Plot age_to_duration distribution vs Target
plt.subplot(1, 2, 2)
sns.boxplot(
    x='target',
    y='age_to_duration',
    data=df,
    hue='target',
    palette='Set2',
    legend=False,
)
plt.title('Age to Duration Ratio vs Target')
plt.xlabel('Credit Risk Target')
plt.ylabel('Age to Duration Ratio')

plt.tight_layout()
plt.show()
```

```python

# 1. Ordinal Encoding for features with inherent order
ordinal_mappings = {
    'checking_account': {'A14': 0, 'A11': 1, 'A12': 2, 'A13': 3},
    'savings_account': {'A65': 0, 'A61': 1, 'A62': 2, 'A63': 3, 'A64': 4}
}

# Apply Ordinal Mapping if categorical text columns exist
for col, mapping in ordinal_mappings.items():
  if col in df.columns and df[col].dtype in ['object', 'string']:
    df[col] = df[col].map(mapping)

# 2. Select nominal categorical columns explicitly (Pandas 3.0+ compatible)
nominal_cols = [
    col
    for col in df.columns
    if df[col].dtype in ['object', 'string', 'category']
]

# 3. Apply One-Hot Encoding with drop_first=True to avoid dummy variable trap
df_encoded = pd.get_dummies(df, columns=nominal_cols, drop_first=True)

print("--- Data Shape After Encoding ---")
print(f"Original Shape: {df.shape}")
print(f"Encoded Shape:  {df_encoded.shape}")
```

Train / Test Split


```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# 1. Separate Target Variable (y) from Features (X)
X = df_encoded.drop(columns=['target'])
y = df_encoded['target']

# 2. Train-Test Split (80% Train, 20% Test) with Stratification
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# 3. Identify Numerical Features for Feature Scaling
# (Scale only continuous and log-transformed features, excluding binary 0/1 columns)
continuous_cols = [
    'credit_amount_log',
    'duration_months_log',
    'age_log',
    'monthly_payment_log',
    'age_to_duration',
]

# 4. Apply StandardScaler (Fit on Train, Transform on both Train & Test)
scaler = StandardScaler()
X_train[continuous_cols] = scaler.fit_transform(X_train[continuous_cols])
X_test[continuous_cols] = scaler.transform(X_test[continuous_cols])

print("--- Preprocessing & Data Split Completed ---")
print(f"X_train Shape: {X_train.shape}")
print(f"X_test Shape:  {X_test.shape}")
print(f"y_train Target Distribution:\n{y_train.value_counts()}")
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.ensemble import GradientBoostingClassifier, RandomForestClassifier
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    precision_score,
    recall_score,
    roc_auc_score,
    roc_curve,
)
from sklearn.preprocessing import StandardScaler

# 1. Clean missing values using SimpleImputer
imputer = SimpleImputer(strategy='median')
X_train_clean = imputer.fit_transform(X_train)
X_test_clean = imputer.transform(X_test)

# 2. Scale features to ensure fast convergence for Logistic Regression
scaler = StandardScaler()
X_train_clean = scaler.fit_transform(X_train_clean)
X_test_clean = scaler.transform(X_test_clean)

# 3. Ensure Target Vector is Binary (0 = Good Risk, 1 = Bad Risk)
if y_train.min() == 1:
  y_train_mapped = y_train.map({1: 0, 2: 1})
  y_test_mapped = y_test.map({1: 0, 2: 1})
else:
  y_train_mapped = y_train
  y_test_mapped = y_test

# 4. Initialize Models with increased iterations for Logistic Regression
models = {
    'Logistic Regression': LogisticRegression(random_state=42, max_iter=2000),
    'Random Forest': RandomForestClassifier(random_state=42, n_estimators=100),
    'Gradient Boosting': GradientBoostingClassifier(random_state=42),
}

# 5. Train and Evaluate Each Model
results = []
plt.figure(figsize=(8, 6))

for name, model in models.items():
  # Fit Model
  model.fit(X_train_clean, y_train_mapped)

  # Predictions
  y_pred = model.predict(X_test_clean)
  y_proba = model.predict_proba(X_test_clean)[:, 1]

  # Calculate Metrics
  acc = accuracy_score(y_test_mapped, y_pred)
  rec = recall_score(y_test_mapped, y_pred)
  prec = precision_score(y_test_mapped, y_pred)
  f1 = f1_score(y_test_mapped, y_pred)
  auc = roc_auc_score(y_test_mapped, y_proba)

  results.append({
      'Model': name,
      'Accuracy': round(acc, 4),
      'Recall (Bad Risk)': round(rec, 4),
      'Precision': round(prec, 4),
      'F1-Score': round(f1, 4),
      'ROC-AUC': round(auc, 4),
  })

  # Plot ROC Curve
  fpr, tpr, _ = roc_curve(y_test_mapped, y_proba)
  plt.plot(fpr, tpr, label=f'{name} (AUC = {auc:.3f})')

# 6. Format ROC Plot
plt.plot([0, 1], [0, 1], 'k--', label='Random Chance')
plt.xlabel('False Positive Rate (FPR)')
plt.ylabel('True Positive Rate (TPR / Recall)')
plt.title('ROC Curves Comparison for Credit Risk Models')
plt.legend(loc='lower right')
plt.grid(True, alpha=0.3)
plt.show()

# 7. Display Performance Comparison Table
results_df = pd.DataFrame(results)
print("--- Model Performance Comparison ---")
print(results_df.to_string(index=False))
```

```python


# 1. Define Financial Cost Matrix Assumptions (in USD or relative units)
COST_FALSE_NEGATIVE = 5  # Cost of giving credit to a Bad Risk customer (Loss of Principal)
COST_FALSE_POSITIVE = 1  # Cost of rejecting a Good Risk customer (Lost Interest Opportunity)


def calculate_financial_cost(y_true, y_pred):
  # Confusion Matrix components
  fn = np.sum((y_true == 1) & (y_pred == 0))  # Bad customer given credit
  fp = np.sum((y_true == 0) & (y_pred == 1))  # Good customer rejected

  total_cost = (fn * COST_FALSE_NEGATIVE) + (fp * COST_FALSE_POSITIVE)
  return total_cost, fn, fp


# 2. Evaluate Financial Cost for Each Model at Default Threshold (0.5)
cost_results = []

for name, model in models.items():
  y_proba = model.predict_proba(X_test_clean)[:, 1]
  y_pred = (y_proba >= 0.5).astype(int)

  cost, fn, fp = calculate_financial_cost(y_test_mapped, y_pred)

  cost_results.append({
      'Model': name,
      'False Negatives (Bad Approved)': fn,
      'False Positives (Good Rejected)': fp,
      'Total Financial Cost Score': cost,
  })

cost_df = pd.DataFrame(cost_results).sort_values(
    by='Total Financial Cost Score'
)
print("--- Financial Cost Matrix Analysis (Default Threshold = 0.5) ---")
print(cost_df.to_string(index=False))
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. Optimize Threshold for Gradient Boosting Model
gb_model = models['Gradient Boosting']
y_proba_gb = gb_model.predict_proba(X_test_clean)[:, 1]

thresholds = np.linspace(0.05, 0.95, 91)
cost_list = []

for t in thresholds:
    y_pred_t = (y_proba_gb >= t).astype(int)
    fn = np.sum((y_test_mapped == 1) & (y_pred_t == 0))
    fp = np.sum((y_test_mapped == 0) & (y_pred_t == 1))
    total_cost = (fn * 5) + (fp * 1)
    cost_list.append({'Threshold': t, 'Cost': total_cost, 'FN': fn, 'FP': fp})

cost_threshold_df = pd.DataFrame(cost_list)
min_cost_row = cost_threshold_df.loc[cost_threshold_df['Cost'].idxmin()]

# 2. Plot Financial Cost vs Decision Threshold
plt.figure(figsize=(9, 5))
plt.plot(cost_threshold_df['Threshold'], cost_threshold_df['Cost'], color='#d62728', linewidth=2, label='Total Cost Curve')
plt.axvline(x=min_cost_row['Threshold'], color='black', linestyle='--', label=f"Optimal Threshold ({min_cost_row['Threshold']:.2f})")

plt.title('Optimization of Decision Threshold based on Cost Matrix (Gradient Boosting)')
plt.xlabel('Decision Threshold')
plt.ylabel('Total Financial Cost Score')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

print(f"--- Optimal Decision Threshold Results ---")
print(f"Default Threshold Cost (0.50): 161")
print(f"Optimal Threshold: {min_cost_row['Threshold']:.2f}")
print(f"Minimum Financial Cost Score: {int(min_cost_row['Cost'])}")
print(f"Bad Risk Approved (FN): {int(min_cost_row['FN'])} | Good Risk Rejected (FP): {int(min_cost_row['FP'])}")
```
